```python
# encodage de la cible 
data['status']=data['status'].map({'positive':0,'negative':1})
print(data['status'])
```

```python
# matrice de correlation 
import matplotlib.pyplot as plt 
import seaborn as sns
num_cols=['status','age', 'pressurehight', 'pressurelow', 'glucose', 'kcm', 'troponin', 'impluse']
matrice_corre=data[num_cols].corr()
plt.figure(figsize=(8,4))
sns.heatmap(matrice_corre,annot=True, cmap='coolwarm', fmt='.2f')
plt.show()
```
